```python
import numpy as np
```

```python
marks = np.array([55, 72, 68, 90, 45, 88, 76, 60, 95, 50])

condition = marks > 70
print(condition)
```

```python
marks = np.array([55, 72, 68, 90, 45, 88, 76, 60, 95, 50])

passed = marks[(marks > 70) & (marks < 80)]
print(passed)
```

```python
marks = np.array([55, 72, 68, 90, 45, 88, 76, 60, 95, 50])

result = np.where(marks >= 70, "Pass", "Fail")
print(result)
```

```python
arr = np.array([10, -5, 20, -8, 15, -3])

result = np.where(arr < 0, 0, arr)
print(result)
```

```python
marks = np.array([55, 72, 68, 90, 45, 88, 76, 60, 95, 50])

passed_count = np.sum(marks >= 70)
print("Students who passed:", passed_count)

failed_count = np.sum(marks < 70)
print("Students who failed:", failed_count)
```

```python
scores = np.array([[85, 45, 78],
                   [92, 38, 55],
                   [70, 75, 30]])

print(scores[scores > 90])
```
